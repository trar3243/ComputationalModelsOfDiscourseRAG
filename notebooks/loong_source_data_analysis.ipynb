# Data Analysis: Loong Corpus\n\nQuick overview of the three document categories: **financial**, **legal**, and **paper**.

```python
import os, json, random
from pathlib import Path

DATA_DIR = Path("data/doc")

# ── helpers ──────────────────────────────────────────────────────────────────

def word_count(text: str) -> int:
    return len(text.split())

def load_financial():
    docs = []
    for p in sorted((DATA_DIR / "financial").glob("*.txt")):
        text = p.read_text(errors="ignore")
        docs.append({"name": p.name, "text": text})
    return docs

def load_legal():
    data = json.loads((DATA_DIR / "legal" / "legal.json").read_text())
    docs = []
    for title, meta in data.items():
        text = meta.get("content", "")
        docs.append({"name": title, "text": text, "meta": meta})
    return docs

def load_paper():
    docs = []
    for p in sorted((DATA_DIR / "paper").glob("*.md")):
        text = p.read_text(errors="ignore")
        docs.append({"name": p.name, "text": text})
    return docs

financial = load_financial()
legal     = load_legal()
paper     = load_paper()

print(f"Loaded  financial: {len(financial):>4} docs")
print(f"Loaded  legal:     {len(legal):>4} docs")
print(f"Loaded  paper:     {len(paper):>4} docs")
```

## Doc counts & word-count statistics

```python
import statistics

def summarise(name, docs):
    wcs = [word_count(d["text"]) for d in docs]
    print(f"{'─'*50}")
    print(f"  Category : {name}")
    print(f"  Num docs : {len(docs)}")
    print(f"  Mean wc  : {statistics.mean(wcs):,.0f}")
    print(f"  Median wc: {statistics.median(wcs):,.0f}")
    print(f"  Min wc   : {min(wcs):,}")
    print(f"  Max wc   : {max(wcs):,}")

summarise("financial", financial)
summarise("legal",     legal)
summarise("paper",     paper)
```

## Text samples (3 random docs per category, first 300 words each)

```python
random.seed(42)

def show_samples(name, docs, n=3, preview_words=300):
    print(f"\n{'═'*60}")
    print(f"  {name.upper()} — {n} random samples")
    print(f"{'═'*60}")
    sample = random.sample(docs, min(n, len(docs)))
    for i, doc in enumerate(sample, 1):
        snippet = " ".join(doc["text"].split()[:preview_words])
        print(f"\n[{i}] {doc['name']}")
        print(f"    word count: {word_count(doc['text']):,}")
        print(f"\n{snippet}{'…' if word_count(doc['text']) > preview_words else ''}")
        print()

show_samples("financial", financial)
show_samples("legal",     legal)
show_samples("paper",     paper)
```
