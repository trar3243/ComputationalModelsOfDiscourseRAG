# Document Repository Counts
Counts usable documents (non-empty, valid files) in each subdirectory of `documents/`.

```python
from pathlib import Path
import pandas as pd

DOCS_DIR = Path('documents')
EXTENSIONS = {'.txt', '.md', '.json', '.jsonl'}

# Repos: top-level files + each immediate subdirectory of loong_docs
repos = {
    'state_of_the_union': DOCS_DIR,
    'financial':          DOCS_DIR / 'loong_docs' / 'financial',
    'paper':              DOCS_DIR / 'loong_docs' / 'paper',
}

def count_docs(path: Path, top_level_only: bool = False) -> int:
    """Count non-empty files with a usable extension."""
    files = path.iterdir() if top_level_only else path.rglob('*')
    return sum(
        1 for f in files
        if f.is_file()
        and f.suffix in EXTENSIONS
        and f.stat().st_size > 0
    )

rows = [
    {
        'repo':      name,
        'path':      str(path),
        'doc_count': count_docs(path, top_level_only=(name == 'state_of_the_union')),
    }
    for name, path in repos.items()
]

df = pd.DataFrame(rows)
df.loc[len(df)] = ['TOTAL', '', df['doc_count'].sum()]
display(df.to_string(index=False))
```
